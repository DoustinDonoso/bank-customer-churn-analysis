```python

import pandas as pd

df = pd.read_csv("Customer-Churn-Records-clean.csv")

# --- Tasa de abandono general ---
print("=== CHURN GENERAL ===")
print("n:", len(df))
print("Exited=1:", int(df["Exited"].sum()))
print("Churn rate:", df["Exited"].mean())
print(df["Exited"].value_counts(normalize=True))

# --- Demografía: Geography, Gender ---
print("\n=== CHURN POR GEOGRAPHY ===")
geo = (
    df.groupby("Geography", as_index=False)
      .agg(n=("Exited", "size"), churners=("Exited", "sum"), churn_rate=("Exited", "mean"))
)
print(geo.sort_values("churn_rate", ascending=False).to_string(index=False))

print("\n=== CHURN POR GENDER ===")
gen = (
    df.groupby("Gender", as_index=False)
      .agg(n=("Exited", "size"), churners=("Exited", "sum"), churn_rate=("Exited", "mean"))
)
print(gen.sort_values("churn_rate", ascending=False).to_string(index=False))

print("\n=== CHURN POR GEOGRAPHY x GENDER ===")
gg = (
    df.groupby(["Geography", "Gender"], as_index=False)
      .agg(n=("Exited", "size"), churners=("Exited", "sum"), churn_rate=("Exited", "mean"))
)
print(gg.sort_values(["Geography", "Gender"]).to_string(index=False))

# --- CreditScore ---
print("\n=== CREDITSCORE vs EXITED ===")
print(df.groupby("Exited")["CreditScore"].describe())
df["CreditScoreBand"] = pd.cut(
    df["CreditScore"],
    bins=[349, 500, 600, 700, 800, 850],
    labels=["350-500", "501-600", "601-700", "701-800", "801-850"],
)
cs = (
    df.groupby("CreditScoreBand", observed=True, as_index=False)
      .agg(n=("Exited", "size"), churners=("Exited", "sum"), churn_rate=("Exited", "mean"))
)
print(cs.to_string(index=False))

# --- Balance ---
print("\n=== BALANCE vs EXITED ===")
print(df.groupby("Exited")["Balance"].describe())
print("Churn Balance==0:", df.loc[df["Balance"] == 0, "Exited"].mean(), "n=", (df["Balance"] == 0).sum())
print("Churn Balance>0:", df.loc[df["Balance"] > 0, "Exited"].mean(), "n=", (df["Balance"] > 0).sum())
if "BalanceCategory" in df.columns:
    bc = (
        df.groupby("BalanceCategory", as_index=False)
          .agg(n=("Exited", "size"), churners=("Exited", "sum"), churn_rate=("Exited", "mean"))
    )
    print(bc.sort_values("churn_rate", ascending=False).to_string(index=False))

# --- NumOfProducts ---
print("\n=== NUMOFPRODUCTS vs EXITED ===")
np_ = (
    df.groupby("NumOfProducts", as_index=False)
      .agg(n=("Exited", "size"), churners=("Exited", "sum"), churn_rate=("Exited", "mean"))
)
print(np_.to_string(index=False))
print(pd.crosstab(df["NumOfProducts"], df["Exited"], margins=True))

# --- Complain ---
print("\n=== COMPLAIN vs EXITED ===")
print(pd.crosstab(df["Complain"], df["Exited"], margins=True))
print(pd.crosstab(df["Complain"], df["Exited"], normalize="index"))
print("Churn por Complain:")
print(df.groupby("Complain")["Exited"].mean())
print("Discrepancias Complain != Exited:", int((df["Complain"] != df["Exited"]).sum()))

# --- Satisfaction Score ---
print("\n=== SATISFACTION SCORE vs EXITED ===")
sat = (
    df.groupby("Satisfaction Score", as_index=False)
      .agg(n=("Exited", "size"), churners=("Exited", "sum"), churn_rate=("Exited", "mean"))
)
print(sat.to_string(index=False))
print(df.groupby("Exited")["Satisfaction Score"].describe())

print("\n=== COMPLAIN x SATISFACTION (churn rate) ===")
print(
    df.pivot_table(
        index="Satisfaction Score",
        columns="Complain",
        values="Exited",
        aggfunc="mean",
    )
)

# --- IsActiveMember ---
print("\n=== ISACTIVEMEMBER vs EXITED ===")
act = (
    df.groupby("IsActiveMember", as_index=False)
      .agg(n=("Exited", "size"), churners=("Exited", "sum"), churn_rate=("Exited", "mean"))
)
print(act.to_string(index=False))
print(pd.crosstab(df["IsActiveMember"], df["Exited"], margins=True))
print(pd.crosstab(df["IsActiveMember"], df["Exited"], normalize="index"))

print("\n=== ISACTIVEMEMBER x GEOGRAPHY (churn rate) ===")
print(
    df.pivot_table(
        index="Geography",
        columns="IsActiveMember",
        values="Exited",
        aggfunc="mean",
    )
)

print("\n=== ISACTIVEMEMBER x NUMOFPRODUCTS (churn rate) ===")
print(
    df.pivot_table(
        index="NumOfProducts",
        columns="IsActiveMember",
        values="Exited",
        aggfunc=["mean", "size"],
    )
)
```
